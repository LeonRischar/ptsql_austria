```python
import osmnx as ox
import pickle
import time
```

```python
PATH_GRAPHS_IN = "../data/in/osmnx_graphs/"
```

```python
# function to convert graphml to pkl
# this should speed up loading the graphs

def convert_graphml_to_pkl(file_names=[]):
    for name in file_names:
        n = "graph_" + name
        print("Loading graph: " + n)
        start_time = time.time()
        g = ox.load_graphml(f"{PATH_GRAPHS_IN}{n}.graphml")
        load_time = time.time()
        print(f"Loaded {n} in {load_time - start_time:.3f}s")

        with open(f"{PATH_GRAPHS_IN}{n}.pkl", "wb") as f:
            pickle.dump(g, f, protocol=pickle.HIGHEST_PROTOCOL)

        print(f"Saved {n} to {PATH_GRAPHS_IN}{n}.pkl")

```

```python
names = ["all_regions"]#["Vienna", "Lower Austria", "Upper Austria", "Burgenland", "Salzburg", "Styria", "Carinthia", "Tyrol", "Vorarlberg"] #"all_regions", 
```

```python
convert_graphml_to_pkl(names)
```

```python
def load_pickle_graph(name):
    n = "graph_" + name
    print(f"Loading graph {n}")
    start_time = time.time()
    with open(f"{PATH_GRAPHS_IN}{n}.pkl", "rb") as f:
        g = pickle.load(f)
    load_time = time.time()
    print(f"Loaded graph {n} in {load_time - start_time:.3f}s")

    print(len(g.nodes))
    print(len(g.edges))

    return g
```

```python
load_pickle_graph("all_regions")
```
